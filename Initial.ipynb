```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
tickers = ["SPY", "QQQ", "NVDA", "JPM", "TLT", "GLD"]
```

```python
weights = {
    "SPY": 0.3,
    "QQQ": 0.2,
    "NVDA": 0.15,
    "JPM": 0.10,
    "TLT": 0.15,
    "GLD": 0.10
}
sum_weights = sum(weights.values())
print(sum_weights)
```

```python
prices = yf.download(
    tickers,
    start = "2021-01-01",
    end = "2026-10-01",
    auto_adjust = True
)
prices.head
prices.columns
close_prices = prices['Close']
close_prices.head()
```

```python
close_prices.shape
close_prices.info()
close_prices.isna().sum()
close_prices = close_prices[
    ["SPY", "QQQ", "NVDA", "JPM", "TLT", "GLD"]
]
```

```python
close_prices.plot(figsize=(12, 6))

plt.title("Historical Asset Prices")
plt.xlabel("Date")
plt.ylabel("Adjusted Price")
plt.show()
```

```python
close_prices.to_csv("Data/Asset_prices.csv")
```
